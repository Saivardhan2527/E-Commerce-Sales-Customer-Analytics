```python
# customer analysis
customer_analysis = (
    orders.groupby("customer_id")
    .agg(
        orders=("order_id", "nunique"),
        revenue=("revenue", "sum"),
        profit=("profit", "sum"),
        last_order=("order_date", "max")
    )
)

customer_analysis.head()
```

```python
# repeat customer analysis
customer_type = customer_analysis.copy()

customer_type["customer_type"] = np.where(
    customer_type["orders"] == 1,
    "One-Time Customer",
    "Repeat Customer"
)

customer_type["customer_type"].value_counts()
```

```python
# product analysis
product_analysis = orders.groupby("product_id").agg(tot_profit = ("profit","sum"),
                                                    tot_reveneue = ("revenue","sum"),
                                                    tot_quantity = ("quantity","sum"))
product_analysis = product_analysis.merge(products[["product_id","category","subcategory","product_name"]],  on = "product_id", how="left")
product_analysis
```

```python
# rfm analysis

rfm = (
    orders.groupby("customer_id")
    .agg(
        recency=("order_date", "max"),
        frequency=("order_id", "nunique"),
        monetary=("revenue", "sum")
    )
    .reset_index()
)
rfm
```

```python
x = pd.Timestamp("2025-12-31")

rfm["recency"] = (x-rfm["recency"]).dt.days
rfm
```

```python
rfm["r_score"] = pd.qcut(
    rfm["recency"].rank(method="first"),
    5,
    labels=[5, 4, 3, 2, 1]
).astype(int)
```

```python
rfm["f_score"] = pd.qcut(
    rfm["frequency"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)
```

```python
rfm["m_score"] = pd.qcut(
    rfm["monetary"].rank(method="first"),
    5,
    labels=[1, 2, 3, 4, 5]
).astype(int)
```

```python
rfm["rfm_score"] = rfm["r_score"] + rfm["f_score"] + rfm["m_score"]
rfm
```

```python
def segment_customer(score):
    if score >= 13:
        return "High Value"
    elif score >= 10:
        return "Medium Value"
    elif score >= 7:
        return "Low Value"
    else:
        return "At Risk"
```

```python
rfm["segment"] = rfm["rfm_score"].apply(segment_customer)
rfm
```

```python
# revenue by segment

segment_analysis = rfm.groupby("segment").agg(rfm_score = ("rfm_score","sum"),
                                              revenue = ("monetary","sum"),
                                              avg_revenue = ("monetary","mean"))
segment_analysis
```
